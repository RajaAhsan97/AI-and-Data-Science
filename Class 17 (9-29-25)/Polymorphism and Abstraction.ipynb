Polymorphism:  
Polymorphism is a way in which different classes can share the same method name, but those methods can act differently based on which objects call them.
In Polymorphism, the methods of classes i.e. Parent/s or child/s are same, but the underlying execution differs depending on the object’s class.


```python
# Parent class or Base Class or Super class
class Animal:
    # constructor
    def __init__(self, name):
        # Instance Attribute
        self.name = name

    # Instance Method
    def speak(self):
        return f"All animals makes sound"

# Child Class 1
class Cat(Animal):
    # Instance Method
    def speak(self):
        return f"{self.name} says Meow"

# Child Class 2
class Dog(Animal):
    # Instance Method
    def speak(self):
        return f"{self.name} says Woof"

# Using Polymorphism
animal = Animal("pets")
print(type(animal))
print(animal.speak())

cat = Cat("Tom")
print(type(cat))
print(cat.speak())

dog = Dog("Tuffy")
print(type(dog))
print(dog.speak())
```

Abstraction:  
Abstraction means hiding implementation details and showing only the essential features to the users.  
In Python, Abstraction is implemented using abstract classes and abstract methods.
The concept of hiding the complex reality while exposing only the essential functionality or interfaces to the users. It helps reduce programming complexity and efforts.  
We cannot create the object of abstract class directly. It’s purpose is to be directly inherited by other classes.
Abstract classes defines a set of methods that subclasses are required to implement (Blueprint for classes)


Class Abstraction  
1. Python provides built-in module for abstract class i.e. abc (Abstract base classes)  
2. for making abstract class we must make atleast one method an abstract method  
3. For making abstract method we must use decorator method  
4. We cannot make object of abstract class directly  
5. An abstract class can have abstract method and normal method.  
6. We can define @abstractmethod in the Parent class  and define its implementations in the child class.


```python
from abc import ABC, abstractmethod
```

```python
class Vehicle(ABC):
    def __init__(self, no_of_tyres):
        # Instance attributes
        self.no_of_tyres = no_of_tyres

    # Abstract method
    @abstractmethod
    def start(self):
        pass

    # normal method
    def display(self):
        return f"This is vehicle"

class Bike(Vehicle):
    # constructor
    def __init__(self, no_of_tyres, engine):
        super().__init__(no_of_tyres)

        self.engine_type = engine

    # Instance Method
    def start(self):
        return f"start with kick, engine type {self.engine_type}"

class Scooty(Vehicle):
    # constructor
    def __init__(self, no_of_tyres, engine, gears):
        super().__init__(no_of_tyres)
        self.engine_type = engine
        self.gears = gears

    # Instance Method
    def start(self):
        return f"self start, engine type {self.engine_type}"
```

```python
bike = Bike(2, "Petrol")
print(bike.start())
print(bike.display())

scooty = Scooty(4, "Electric Vehicle", 5)
print(scooty.start())
print(scooty.display())
```
