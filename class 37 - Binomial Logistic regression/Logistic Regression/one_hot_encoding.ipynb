```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, classification_report
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('iris dataset/archive/iris.csv')
```

```python
df
```

```python
df['Species'].unique()
```

```python
X = df.drop(['Id', 'Species'], axis=1)
Y = df[['Species']]  # one hot encoder eccept 2D
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(Y)
```

```python
y_encoded
```

```python
encoder = OneHotEncoder(sparse_output=False)

y_encoded = encoder.fit_transform(Y)
```

```python
y_encoded[:5]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
```

```python
X_train
```

```python
y_train
```

```python
import numpy as np

y_train_labels = np.argmax(y_train, axis=1)
y_test_labels = np.argmax(y_test, axis=1)
```

```python
y_train_labels
```

```python
scalar = StandardScaler()

X_train_scaled = scalar.fit_transform(X_train)
X_test_scaled = scalar.transform(X_test)
```

```python
logistic_reg = LogisticRegression()
```

```python
logistic_reg.fit(X_train_scaled, y_train_labels)
```

```python
y_test_pred = logistic_reg.predict(X_test_scaled)
```

```python
conf_mat = confusion_matrix(y_test_labels, y_test_pred)
```

```python
import seaborn as sns
```

```python
sns.heatmap(conf_mat, annot=True)
```

```python
print(classification_report(y_test_labels, y_test_pred))
```
