# One-Hot Encoding + Logistic Regression

This notebook demonstrates:
- Loading a categorical dataset
- Applying One-Hot Encoding on independent features
- Training a Logistic Regression model

## 1. Load Dataset

```python
import pandas as pd
```

```python
df = pd.read_csv('titanic dataset/titanic.csv')
```

```python
df
```

## 2. Separate Features and Target

```python

X = df.drop("income", axis=1)
y = df["income"]
```

## 3. One-Hot Encoding (Independent Features Only)

```python

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

categorical_cols = X.select_dtypes(include="object").columns
numerical_cols = X.select_dtypes(exclude="object").columns

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(drop="first", sparse_output=False), categorical_cols),
        ("num", "passthrough", numerical_cols)
    ]
)

X_encoded = preprocessor.fit_transform(X)
```

## 4. Train Logistic Regression

```python

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
accuracy_score(y_test, y_pred)
```

## 5. Conclusion
One-hot encoding is applied only to categorical independent features before training Logistic Regression.
