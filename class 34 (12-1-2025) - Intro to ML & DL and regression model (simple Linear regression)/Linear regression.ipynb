## Machine Learning vs Deep Learning

#### MACHINE LEARNING: 
1. In ML, model learns patterns from data and adjusts its parameters to make predictions. 
2. We need to perform feature engineering (i.e. extract relavent features on which output depends)
3. Machine Learning models requires small data
4. ML algorithms easily resolves tabular data

#### DEEP LEARNING: 
1. DL works similarly, but it uses multi-layer neural networks that update their weights through backpropagation whenever the prediction is incorrect.
2. In Deep learning -> the algorithm automatically filters the irrelavent features.
3. Deep learning models require large amount of data
4. DL algorithms resolves text, speech, images

| Aspect              | Machine Learning   | Deep Learning              |
| ------------------- | ------------------ | -------------------------- |
| Part of             | AI                 | ML                         |
| Data Required       | Low to medium      | High                       |
| Feature Engineering | Manual             | Automatic                  |
| Training Time       | Fast               | Slow                       |
| Accuracy            | Good               | Often best (on large data) |
| Hardware            | CPU is enough      | Requires GPU               |
| Transparency        | More interpretable | Black-box model            |

Y = w<sub>1</sub>X<sub>1</sub> + c

### Let's try this simple model

```python
def linear_model(x, w=0, c=0):
    y = w*x + c
    return y
```

#### Cost Function

It is used to measure the performance of the model. The cost or error function is used to guide the model of finding the best parameter by minimizing the errors between predicted and actual outcomes.
The lower the cost function values, the better the performance of the model

#### Mean Squared Error (MSE)
The average of the Squares of difference between the actual and predicted outcomes

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$


### Root Mean Squared Error (RMSE)

$$
RMSE = \sqrt{ \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y_i})^2 }
$$


### Mean Absolute Error (MAE)

$$
MAE = \frac{1}{n} \sum_{i=1}^{n} \left| y_i - \hat{y}_i \right|
$$


```python
def mean_square_error(y_true, y_pred):
    errors = (y_true - y_pred) ** 2
    mse = errors.mean()
    return mse
```

```python
X = df[['Height']].values
y_true = df[['Weight']].values
```

```python
y_pred = linear_model(X, 7.7, -349.7)        # y = mx + c

cost_function = mean_square_error(y_true, y_pred) ** 0.5

print("Error: ", cost_function)

df.plot(x='Height', y='Weight', kind='scatter', title='Weight vs Height')

plt.plot([55,80], [60, 275], color='red', linewidth=2)

plt.plot(X, y_pred, color='green', linewidth=2)

plt.show()
```

### Machine Learning using scikit learn

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
lr_model = LinearRegression()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_true, test_size=0.2, random_state=42)
```

```python
lr_model.fit(X_train, y_train)
```

```python
# checking model parameters

print("Coefficient: ", lr_model.coef_)
print("intercept: ", lr_model.intercept_)
```

#### Evaluate model Performance - on Training Data

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error
```

```python
y_pred = lr_model.predict(X_train)

print("Mean Squared Error (MSE): ", mean_squared_error(y_train, y_pred))
print("Root Mean Squared Error (RMSE): ", root_mean_squared_error(y_train, y_pred))
print("Mean Absolute Error (MAE): ", mean_absolute_error(y_train, y_pred))
```

```python
print("R2 score: ", r2_score(y_train, y_pred))
```

#### Evaluating model performance - on Test data

```python
y_pred = lr_model.predict(X_test)

print("MSE: ", mean_squared_error(y_test, y_pred))
print("RMSE: ", root_mean_squared_error(y_test, y_pred))
print("MAE: ", mean_absolute_error(y_test, y_pred))
print("R2 score: ", r2_score(y_test, y_pred))
```
