# 2. Removing unimportant things

a. Remove HTML tags

```python
import re
```

```python
def remove_html_tags(text):
  patterns = re.compile('<.*?>')
  return patterns.sub(r'',text)
```

```python
text = '<p>This movie was <b>absolutely amazing</b>! I loved the storyline and the acting.<br /><br />The cinematography was stunning and the soundtrack was beautiful.<br />Rating: 9/10</p>'
```

```python
remove_html_tags(text)
```

b. Remove urls

```python
def remove_url(text):
  pattern = re.compile(r'https?://|S+|www\.\S+')
  return pattern.sub(r'', text)
```

```python
text1 = 'I checked the movie details on https://www.imdb.com before watching it'
text2 = 'Out official website is https://www.saylanimit.com. Visit https://www.saylanimit.com for details'
```

```python
remove_url(text1)
```

```python
remove_url(text2)
```

```python
df['review'] = df['review'].apply(remove_url)
```

c. Remove Punctuations

```python
import string
```

```python
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punctuation(text):
  for char in exclude:
    text = text.replace(char, '')
  return text
```

```python
text = 'string, with, punctuations?'
```

```python
remove_punctuation(text)
```

```python
def remove_punctuation1(text):
  return text.translate(str.maketrans('', '', exclude))
```

```python
remove_punctuation1(text)
```

d. Remove stopwords

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
  new_text = []

  for word in text.split():
    if word in stopwords.words('english'):
      new_text.append('')
    else:
      new_text.append(word)

  return " ".join(new_text)
```

```python
text1 = 'This is the sample text.'
text2 = 'You are very lucky'
```

```python
remove_stopwords(text1)
```

```python
remove_stopwords(text2)
```

# Inflection:
means a same word is represented in different forms

walk, walking, walked, walk!


# Stemming:
is the process of reducing inflection

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem_words(text):
  return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = 'walk walking walked walks'
```

```python
stem_words(sample)
```

```python
text = 'The runner was running quickly because he wanted to win the race. He studied the track, seeing the differences in the jumps.'
```

```python
stem_words(text)
```

# Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
```

```python
wordnet_lemmatizer = WordNetLemmatizer()
```

```python
text = 'The runner was running quickly because he wanted to win the race. He studied the track, seeing the differences in the jumps.'
```

```python
sentence_word = nltk.tokenize.word_tokenize(text)
```

```python
sentence_word
```

```python
punctuations = string.punctuation
```

```python
punctuations
```

```python
for word in sentence_word:
  if word in punctuations:
    sentence_word.remove(word)
```

```python
sentence_word
```

```python
print('{0:20}{1:20}'.format("word", "Lemma"))
for word in sentence_word:
  print('{0:20}{1:20}'.format(word, wordnet_lemmatizer.lemmatize(word, pos='v')))
```
