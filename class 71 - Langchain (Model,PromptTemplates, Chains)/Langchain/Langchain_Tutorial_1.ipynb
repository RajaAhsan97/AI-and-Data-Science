```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
```

```python
from google.colab import userdata

api_key = userdata.get('OPENROUTER_SECRET')
```

```python
llm = ChatOpenAI(
    api_key=api_key,
    model="nvidia/nemotron-3-nano-omni-30b-a3b-reasoning:free",
    base_url="https://openrouter.ai/api/v1"
)
```

```python
response = llm.invoke([HumanMessage(content="Hi how are you?")])

print(response)
```

```python
response.content
```

# Simple Prompt (Static prompt)

```python
response = llm.invoke([HumanMessage(content="Explain Langchain in Urdu")])

print(response.content)
```

# Dynamic Prompt
# Prompt Template


```python
from langchain_core.prompts import PromptTemplate
```

```python
template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple words"
)

prompt = template.format(topic="AI")
```

```python
response = llm.invoke(prompt)

print(response.content)
```

# ChatPrompt Template

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {profession} assistant."),
    ("human", "Can you help me with {task}?")
])
```

```python
messages = chat_prompt.format_messages(
    profession="financial",
    task="budget planning"
)
```

```python
response = llm.invoke(messages)

print(response.content)
```
