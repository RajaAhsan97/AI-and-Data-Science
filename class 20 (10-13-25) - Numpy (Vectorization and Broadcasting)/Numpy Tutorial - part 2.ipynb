
## Vectorization

Vectorization means to perform mathematical operation on entire vectors or matrices at once, instead of looping through elements

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([2,3,4,5,6])
```

```python
# element-wise operations (addition, subtraction, multiply, divide)
add_arr = arr1 + arr2
print(add_arr)
sub_arr = arr1 - arr2
print(sub_arr)
mul_arr = arr1 * arr2
print(mul_arr)
div_arr = arr1 / arr2
print(div_arr)
```

```python
arr = np.random.randint(0, 1000, size=(1000,1000))
new_arr = np.sin(arr)
print(new_arr)
```

## Broadcasting   (Explained)

Broadcasting is the ability of Numpy to perform arithmetic operations on arrays of different shapes - by automatically expanding the smaller array so that it matches the shape of the larger one

General broadcasting rules
When operating on two arrays, NumPy compares their shapes element-wise. It starts with the trailing (i.e. rightmost) dimension and works its way left. Two dimensions are compatible when

they are equal, or

one of them is 1.

If these conditions are not met, a ValueError: operands could not be broadcast together exception is thrown, indicating that the arrays have incompatible shapes.

Input arrays do not need to have the same number of dimensions. The resulting array will have the same number of dimensions as the input array with the greatest number of dimensions, where the size of each dimension is the largest size of the corresponding dimension among the input arrays. Note that missing dimensions are assumed to have size one.

### Example 1:  Scalar broadcasting

```python
# Scalar broadcasting

arr = np.array([1,3,5,7])  # shape --> (4,)    or     1x4
# numpy automatically broadcast the scaler 5 to [5,5,5,5] before addition
print(arr + 5)
```

| Step                                   | Description                                                                  | Example in this case   |
| -------------------------------------- | ---------------------------------------------------------------------------- | ---------------------- |
| **1️⃣ Identify shapes**                | `arr` has shape `(4,)`, scalar `5` has shape `()`                            | `(4,)` and `()`        |
| **2️⃣ Align shapes (right to left)**   | NumPy lines them up: <br>`(4,)`<br>`()` → treat scalar as `(1,)`             | `(4,)` and `(1,)`      |
| **3️⃣ Apply compatibility rule**       | Dimensions are **compatible** if: <br>• same value OR <br>• one of them is 1 | `4` vs `1` ✅ compatible     |
| **4️⃣ Expand smaller shape**           | NumPy *virtually* expands scalar `(1,)` → `(4,)`                             | scalar → `[5,5,5,5]`   |
| **5️⃣ Perform element-wise operation** | Adds each element                                                            | `[1+5, 3+5, 5+5, 7+5]` |


### Row Broadcasting:

```python
# Row broadcasting
arr1 = np.array([
    [1,2,3,4,5],
    [11,22,33,44,55]
])      # (2,5)
arr2 = np.array([2,3,4,5,6])   # (1,5)   --> (5,)

new_arr = arr1 + arr2
print(new_arr)

#(2,5)
#(1,5)

#[2,3,4,5,6]
#[2,3,4,5,6]
```

| Step                                  | Description                                                                                              | Example in this case                             |
| ------------------------------------- | -------------------------------------------------------------------------------------------------------- | ------------------------------------------------ |
| **1️⃣ Identify shapes**               | `arr1` → `(2, 5)`  (2 rows × 5 columns)  <br>`arr2` → `(5,)`  (1D array)                                 | `(2, 5)` and `(5,)`                              |
| **2️⃣ Align shapes (right to left)**  | NumPy lines them up like this:<br>`(2, 5)`<br>    `(   5)`                                               | Missing dimension in `arr2` becomes 1 → `(1, 5)` |
| **3️⃣ Apply compatibility rule**      | Compare each dimension **from right to left:** <br>→ `5 vs 5` ✅ (equal) <br>→ `2 vs 1` ✅ (1 can stretch) | ✅ Compatible                                     |
| **4️⃣ Expand smaller array**          | NumPy virtually expands `arr2` into **two rows**: <br>`[[2,3,4,5,6], [2,3,4,5,6]]`                       | Now both `(2,5)`                                 |
| **5️⃣ Perform element-wise addition** | Adds each element positionally                                                                           | Works across both rows                           |


### Column Broadcasting

```python
# column broadcasting
arr1 = np.array([
    [1,2,3,4],
    [2,3,4,5]
])   # (2,4)

arr2 = np.array([
    [2],
    [5]
])  # (2,1)

new_arr = arr1 + arr2
print(new_arr)

#(2,4)
#(2,1)

#[2,2,2,2]
#[5,5,5,5]
```

| Step                                  | Description                                                                                         | Example in this case  |
| ------------------------------------- | --------------------------------------------------------------------------------------------------- | --------------------- |
| **1️⃣ Identify shapes**               | `arr1` → `(2, 4)`  <br>`arr2` → `(2, 1)`                                                            | `(2, 4)` and `(2, 1)` |
| **2️⃣ Align shapes (right to left)**  | Compare each dimension from **right → left**: <br>`4 vs 1` ✅ (1 can stretch) <br>`2 vs 2` ✅ (equal) | ✅ Compatible          |
| **3️⃣ Expand smaller array**          | NumPy **stretches** `arr2` horizontally across 4 columns:  <br>`[[2,2,2,2], [5,5,5,5]]`             | Now both `(2,4)`      |
| **4️⃣ Perform element-wise addition** | Each row of `arr1` adds with its corresponding broadcasted row from `arr2`                          | Works perfectly       |


### Arrays with Incompatible shapes

```python
# Row broadcasting
arr1 = np.array([
    [1,2,3,4,5],
    [11,22,33,44,55]
])    # (2,5)
arr2 = np.array([2,3,4,5])  # (1,4)

new_arr = arr1 + arr2
print(new_arr)

#(2,5)
#(1,4)
```

```python
# Row broadcasting
arr1 = np.array([
    [1,2,3,4,5],
    [11,22,33,44,55]
])
arr2 = np.array([
    [2,3,4,5],
    [1,2,3,4]
])

new_arr = arr1 + arr2
print(new_arr)
```
