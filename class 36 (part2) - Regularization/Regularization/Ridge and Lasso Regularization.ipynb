```python
import mglearn
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
```

```python
X, y = mglearn.datasets.make_wave()
```

```python
X
```

```python
y
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X, y)
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
lr_model = LinearRegression()
```

```python
lr_model.fit(X_train, y_train)
```

```python
print(lr_model.coef_)
print(lr_model.intercept_)
```

```python
print("Training Data: ", r2_score(y_train, lr_model.predict(X_train)))
print("Testing Data: ", r2_score(y_test, lr_model.predict(X_test)))
```

Result: The model underfits

### Overfitting

```python
X, y = mglearn.datasets.load_extended_boston()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

```python
lr_model.fit(X_train, y_train)
```

```python
print("Training Data: ", r2_score(y_train, lr_model.predict(X_train)))
print("Testing Data: ", r2_score(y_test, lr_model.predict(X_test)))
```

### Regularization

#### Ridge (L2) regularization

```python
from sklearn.linear_model import Ridge
```

```python
ridge_model = Ridge(alpha=0.3)
```

```python
ridge_model.fit(X_train, y_train)
```

```python
print("Training Data: ", r2_score(y_train, ridge_model.predict(X_train)))
print("Testing Data: ", r2_score(y_test, ridge_model.predict(X_test)))
```

#### Lasso (L1) Regularization

```python
from sklearn.linear_model import Lasso
```

```python
lasso_model = Lasso(alpha=0.001)
```

```python
lasso_model.fit(X_train, y_train)
```

```python
import numpy as np
```

```python
print("Training Data: ", r2_score(y_train, lasso_model.predict(X_train)))
print("Testing Data: ", r2_score(y_test, lasso_model.predict(X_test)))
print("Number of features used: {}".format(np.sum(lasso_model.coef_ != 0)))
```
