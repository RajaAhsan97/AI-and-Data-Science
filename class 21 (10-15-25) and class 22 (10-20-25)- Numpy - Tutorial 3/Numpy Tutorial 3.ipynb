## Reshape

changes the shape (dimensions) of numpy array without changing it data

```python
# numpy arange() works similar to range() function of Python
# create an array with values are evenly spaced
arr = np.arange(10)    #c start --> 0, end --> 10-1 , step --> 1 (default)

print(arr)
```

```python
arr.shape
```

```python
# Reshaping
new_arr_rshp1 = arr.reshape((2,5))
new_arr_rshp2 = arr.reshape((5,2))
#new_arr = arr.reshape((8,2))
#new_arr = arr.reshape((3,4))  # Error: cannot reshape array of size 10 into shape (3,4)
```

```python
print(new_arr_rshp1)
print(new_arr_rshp1.shape)
```

```python
print(new_arr_rshp2)
print(new_arr_rshp2.shape)
```

reshape() returns a view (not a copy), so changing the reshaped array may affect the original

## Flatten and Ravel  
Flatten --> returns the copy of array collapsed into 1D. Changes to the flatten array not affect the original array

Ravel --> Returns the view of array in collapsed 1D. Changes to raveled array affect the original array


```python
print(new_arr)
# Flatten()
flatten_arr = new_arr.flatten()
print(flatten_arr)
flatten_arr[0] = 100
print(flatten_arr)
print(new_arr)
```

```python
# Ravel()
ravel_arr = new_arr.ravel()
print(ravel_arr)
ravel_arr[0] = 100
print(ravel_arr)
print(new_arr)
```

Array Operations
Numpy support various array operation, which are perfomed elementwise

```python
arr = np.arange(2,19,2, dtype=np.float64)
x = arr.reshape((3,3))
y = np.full((3,3), 3, dtype=np.float64)
print(x)
print(y)
```

```python
# Addition
add = np.add(x,y)
print(add)
# subtraction
sub = np.subtract(x,y)
print(sub)
# multiply 
mul = np.multiply(x,y)
print(mul)
# divide
div = np.divide(x,y)
print(div)
```

Numpy functions
There are also useful builtin functions of Numpy,  by using them we are able to express them across specific axes of the ndarray

axis = 0 --> Combines all rows and gives result per column  
axis = 1 --> Combines all columns and gives result per row

```python
x = np.array([
    [1,2,4],
    [5,6,7]
])

print(np.sum(x))  # Compute sum all elements of x
print(np.sum(x, axis=0))  # Compute sum of each row
print(np.sum(x, axis=1))  # Compute sum of each column
```

```python

print(np.min(x, axis=0))  
print(np.min(x, axis=1))
print(np.max(x, axis=0))
print(np.max(x, axis=1))
```

## argmax() 
compute index of max value in each row

```python
print(np.argmax(x, axis=0))
```

Example: argmax()

```python
students = np.array([
    [85, 92, 78],   # Student 1
    [88, 79, 95],   # Student 2
    [91, 94, 89]    # Student 3
])

# Best subject for each student
best_subjects = np.argmax(students, axis=1)
print("Best subject index for each student:", best_subjects)

# Best student in each subject
best_students = np.argmax(students, axis=0)
print("Best student index for each subject:", best_students)
```

## Filtering

We can also use boolean mask. Suppose we want to all elements of array with value greater tha MAX to MAX

```python
MAX = 5
x = np.array([1, 4, 10, -1, 15, 0, 5])
print(x > MAX)    # returns an array of TRUE and FALSE base on the condition

x[x > MAX] = MAX
print(x)
```

## Stacking

Joins a sequence of an arrays along new axis

```python
a = np.arange(9).reshape(3,3)
print(a)
```

```python
b = np.arange(10,19).reshape(3,3)
print(b)
```

```python
c = np.arange(20,29).reshape(3,3)
print(c)
```

### Horizontal stacking

```python
h_stack = np.hstack((a,b,c))
print(h_stack)
```

### Vertical Stacking

```python
v_stack = np.vstack((a, b, c))
print(v_stack)
```

```python
d = np.arange(3)
print(d)
```

```python
v_stack_new = np.vstack((v_stack, d))
print(v_stack_new)
```
