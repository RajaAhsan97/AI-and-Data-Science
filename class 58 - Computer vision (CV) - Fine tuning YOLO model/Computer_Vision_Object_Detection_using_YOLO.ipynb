```python
def predict(image):
  results = trained_model.predict(image, conf=0.5)

  # for annotating image
  # 1. Use built in annotation
  #annotated_frame = results[0].plot()

  # or

  # 2. Use Custom annotation
  # Draw detections
  class_names = trained_model.names
  annotated_frame = image.copy()
  for r in results:
    boxes = r.boxes.xyxy.cpu().numpy()
    scores = r.boxes.conf.cpu().numpy()
    classes = r.boxes.cls.cpu().numpy().astype(int)

    for box, score, cls in zip(boxes, scores, classes):
      x1, y1, x2, y2 = map(int, box)

      label = f"{class_names[cls]} {score:.2f}"

      # Bounding box
      cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

      # Label background
      (w, h), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2)
      cv2.rectangle(annotated_frame, (x1, y1 - h - 10), (x1 + w, y1), (0, 255, 0), -1)

      # Label text
      cv2.putText(annotated_frame, label, (x1, y1 - 5),
                  cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 2)


  return annotated_frame

```

```python
interface = gr.Interface(fn=predict,
                         inputs=gr.Image(type='numpy'),
                         outputs=gr.Image(type='numpy'),
                         title="YOLO - License plate detection")
interface.launch()
```

```python
def process_video(input_video_path):
  # Input video (0 for webcam)
  cap = cv2.VideoCapture(input_video_path)

  # Get video properties
  frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
  frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
  fps = int(cap.get(cv2.CAP_PROP_FPS))

  # Define VideoWriter (output file)
  output_path = "output_processed.mp4"
  fourcc = cv2.VideoWriter_fourcc(*"mp4v")  # codec
  out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

  # Class names
  class_names = trained_model.names

  while cap.isOpened():
      ret, frame = cap.read()
      if not ret:
          break

      # Run YOLO inference
      results = trained_model.predict(frame, conf=0.5)

      # Draw detections
      for r in results:
          boxes = r.boxes.xyxy.cpu().numpy()
          scores = r.boxes.conf.cpu().numpy()
          classes = r.boxes.cls.cpu().numpy().astype(int)

          for box, score, cls in zip(boxes, scores, classes):
              x1, y1, x2, y2 = map(int, box)

              label = f"{class_names[cls]} {score:.2f}"

              # Bounding box
              cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

              # Label background
              (w, h), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2)
              cv2.rectangle(frame, (x1, y1 - h - 10), (x1 + w, y1), (0, 255, 0), -1)

              # Label text
              cv2.putText(frame, label, (x1, y1 - 5),
                          cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 2)

      # Write frame to output video
      out.write(frame)


  # Release everything
  cap.release()
  out.release()

  return output_path

```

```python
# Gradio Interface
demo = gr.Interface(
    fn=process_video,
    inputs=gr.Video(label="Input Video"),
    outputs=gr.Video(label="Output Video")
)

demo.launch()
```

The next level is combining YOLO license plate detection with OCR to automatically extract the license plate number from video/images

### Pipeline  

Video   
→ YOLO (detect plates)  
→ Crop plates  
→ OCR (read plate number)  
→ Annotate frame + save video

```python
import easyocr
```

```python
import cv2

# Input video (0 for webcam)
cap = cv2.VideoCapture("/istockphoto-1327510104-640_adpp_is.mp4")

## Easy OCR
ocr_reader = easyocr.Reader(['en'])  # English OCR

# Get video properties
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Define VideoWriter (output file)
fourcc = cv2.VideoWriter_fourcc(*"mp4v")  # codec
out = cv2.VideoWriter("/content/licenseplate_and_text_extract.mp4", fourcc, fps, (frame_width, frame_height))

# Class names
class_names = trained_model.names

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Run YOLO inference
    results = trained_model.predict(frame, conf=0.5)

    # Draw detections
    for r in results:
        boxes = r.boxes.xyxy.cpu().numpy()
        scores = r.boxes.conf.cpu().numpy()
        classes = r.boxes.cls.cpu().numpy().astype(int)

        for box, score, cls in zip(boxes, scores, classes):
            x1, y1, x2, y2 = map(int, box)

            # Crop license plate
            plate_img = frame[y1:y2, x1:x2]

            # OCR
            ocr_result = ocr_reader.readtext(plate_img)
            plate_text = " ".join([res[1] for res in ocr_result]) if ocr_result else "N/A"

            label = f"{plate_text} {score:.2f}"

            # Bounding box
            cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Label background
            (w, h), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2)
            cv2.rectangle(frame, (x1, y1 - h - 10), (x1 + w, y1), (0, 255, 0), -1)

            # Label text
            cv2.putText(frame, label, (x1, y1 - 5),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 2)

    # Write frame to output video
    out.write(frame)


# Release everything
cap.release()
out.release()
```
